# پروژه تحلیل اکتشافی داده‌ها (EDA): Airbnb Open Data
## دفترچه ۱ — استخراج، شناخت، پاک‌سازی و آماده‌سازی داده‌ها

**هدف:** آماده‌سازی یک نسخه قابل اتکا از داده‌ها برای تحلیل آماری. این دفترچه عمداً بین «مشاهده مشکل» و «اصلاح داده» تفاوت می‌گذارد؛ هیچ ردیفی صرفاً به دلیل گران بودن به‌عنوان خطا حذف نمی‌شود.

**پرسش اصلی پروژه:** چه الگوهایی در قیمت و ویژگی‌های آگهی‌های Airbnb دیده می‌شود و کدام متغیرها با قیمت ارتباط دارند؟

> راه‌اندازی: فایل `Airbnb_Open_Data.csv` را در پوشه `data` کنار پوشه `notebooks` قرار دهید. این پروژه با Pandas، NumPy، Matplotlib، Seaborn، SciPy و Scikit-learn ساخته شده است.

```python
from pathlib import Path
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from IPython.display import display

warnings.filterwarnings("ignore", category=FutureWarning)
sns.set_theme(style="whitegrid", context="notebook")
pd.set_option("display.max_columns", 50)
pd.set_option("display.max_colwidth", 100)

# مسیر نسبی باعث می‌شود نوت‌بوک پس از استخراج بسته نیز قابل اجرا باشد.
DATA_PATH = Path("../data/Airbnb_Open_Data.csv")
if not DATA_PATH.exists():
    # مسیر جایگزین برای اجرای نوت‌بوک از ریشه پروژه
    DATA_PATH = Path("data/Airbnb_Open_Data.csv")
if not DATA_PATH.exists():
    raise FileNotFoundError("فایل CSV پیدا نشد. آن را در پوشه data قرار دهید.")

df_raw = pd.read_csv(DATA_PATH, low_memory=False)
print(f"تعداد ردیف‌ها: {df_raw.shape[0]:,}")
print(f"تعداد ستون‌ها: {df_raw.shape[1]}")
display(df_raw.head())
```

## ۱. استخراج و شناخت اولیه داده‌ها

در این مرحله ساختار جدول، نام ستون‌ها، نوع داده‌ها و چند ردیف نمونه را بررسی می‌کنیم. این بررسی مشخص می‌کند کدام ستون‌ها عددی واقعی‌اند و کدام ستون‌ها مثل قیمت، به شکل متن ذخیره شده‌اند.

```python
print("ابعاد دیتاست:", df_raw.shape)
print("\nنام ستون‌ها:")
print(df_raw.columns.tolist())
print("\nاطلاعات ستون‌ها:")
df_raw.info()
display(df_raw.describe(include="all").T)
```

### تفسیر اولیه
- `price` و `service fee` ممکن است به‌صورت رشته شامل علامت دلار باشند؛ برای تحلیل عددی باید تبدیل شوند.
- ستون‌های شناسه (`id` و `host id`) شناسه‌اند، نه متغیرهای کمی برای همبستگی.
- ستون‌های متنی مانند `NAME` و `house_rules` معمولاً در تحلیل‌های آماری پایه مستقیماً استفاده نمی‌شوند.
- مقدارهای عجیب یا پرت ابتدا علامت‌گذاری می‌شوند؛ حذف آن‌ها بدون بررسی می‌تواند نتیجه را منحرف کند.

## ۲. کیفیت داده‌ها: مقادیر گمشده، تکراری‌ها و مقادیر یکتا

```python
missing = pd.DataFrame({
    "missing_count": df_raw.isna().sum(),
    "missing_percent": (df_raw.isna().mean() * 100).round(2),
}).sort_values("missing_percent", ascending=False)
display(missing)

print("تعداد ردیف‌های کاملاً تکراری:", int(df_raw.duplicated().sum()))
print("تعداد شناسه‌های آگهی تکراری:", int(df_raw["id"].duplicated().sum()))
print("\nتعداد مقادیر یکتا در ستون‌های دسته‌ای:")
display(df_raw.select_dtypes(include="object").nunique().sort_values(ascending=False).to_frame("unique_values"))
```

### تصمیم پاک‌سازی
۱. متن ستون‌ها پاک‌سازی می‌شود و رشته‌های خالی به مقدار گمشده تبدیل می‌شوند.  
۲. ردیف‌های کاملاً تکراری حذف می‌شوند. اگر شناسه آگهی تکراری باشد، یک رکورد نگه داشته می‌شود تا هر آگهی فقط یک بار شمرده شود؛ تعداد حذف‌ها گزارش می‌شود.  
۳. قیمت‌ها به عدد تبدیل می‌شوند؛ قیمت گمشده برای تحلیل قیمت قابل استفاده نیست، اما در فایل پاک‌سازی‌شده باقی می‌ماند.  
۴. مقادیر ناممکن در ستون‌هایی با دامنه مشخص، مانند تعداد شب منفی یا دسترس‌پذیری خارج از بازه ۰ تا ۳۶۵، به‌عنوان داده نامعتبر به `NaN` تبدیل می‌شوند. این کار به معنی حذف کل آگهی نیست.  
۵. داده‌های پرت قیمت با روش IQR علامت‌گذاری می‌شوند، نه اینکه خودکار حذف شوند.

```python
df = df_raw.copy()

# حذف فاصله‌های اضافی در نام ستون‌ها و مقادیر متنی
df.columns = [c.strip() for c in df.columns]
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].map(lambda x: x.strip() if isinstance(x, str) else x)
    df[col] = df[col].replace({"": np.nan, "nan": np.nan, "None": np.nan})

# یکسان‌سازی چند املای آشکارا ناهماهنگ در گروه منطقه؛ فقط مقادیر شناخته‌شده اصلاح می‌شوند.
if "neighbourhood group" in df.columns:
    area_fix = {
        "brookln": "Brooklyn",
        "manhatan": "Manhattan",
    }
    df["neighbourhood group"] = df["neighbourhood group"].replace(area_fix)

rows_before = len(df)
df = df.drop_duplicates()
exact_duplicates_removed = rows_before - len(df)

duplicate_ids_before = int(df["id"].duplicated().sum())
df = df.drop_duplicates(subset="id", keep="first")
duplicate_ids_removed = duplicate_ids_before

# تبدیل قیمت و کارمزد از رشته‌ای مثل '$1,250 ' به عدد
for col in ["price", "service fee"]:
    df[col] = pd.to_numeric(
        df[col].astype("string").str.replace(r"[$,\s]", "", regex=True),
        errors="coerce"
    )

# تبدیل تاریخ آخرین بازبینی
if "last review" in df.columns:
    df["last review"] = pd.to_datetime(df["last review"], errors="coerce")

# تبدیل ستون‌های منطقی که گاهی رشته و گاهی Boolean هستند
def normalize_bool(value):
    if pd.isna(value):
        return np.nan
    if isinstance(value, (bool, np.bool_)):
        return bool(value)
    v = str(value).strip().lower()
    if v in {"true", "yes", "1"}: return True
    if v in {"false", "no", "0"}: return False
    return np.nan

if "instant_bookable" in df.columns:
    df["instant_bookable"] = df["instant_bookable"].map(normalize_bool)

# تبدیل ستون‌های عددی مورد انتظار
numeric_cols = [
    "lat", "long", "Construction year", "minimum nights",
    "number of reviews", "reviews per month", "review rate number",
    "calculated host listings count", "availability 365"
]
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# اصلاح مقادیر خارج از دامنه‌های منطقی
if "minimum nights" in df.columns:
    df.loc[df["minimum nights"] < 1, "minimum nights"] = np.nan
if "number of reviews" in df.columns:
    df.loc[df["number of reviews"] < 0, "number of reviews"] = np.nan
if "reviews per month" in df.columns:
    df.loc[df["reviews per month"] < 0, "reviews per month"] = np.nan
if "review rate number" in df.columns:
    df.loc[~df["review rate number"].between(1, 5), "review rate number"] = np.nan
if "availability 365" in df.columns:
    df.loc[~df["availability 365"].between(0, 365), "availability 365"] = np.nan
if "Construction year" in df.columns:
    df.loc[~df["Construction year"].between(1800, 2026), "Construction year"] = np.nan
if "calculated host listings count" in df.columns:
    df.loc[df["calculated host listings count"] < 1, "calculated host listings count"] = np.nan
if "price" in df.columns:
    df.loc[df["price"] <= 0, "price"] = np.nan
if "service fee" in df.columns:
    df.loc[df["service fee"] < 0, "service fee"] = np.nan

# برای آگهی‌هایی که تعداد نظر صفر است، نبودن نرخ نظر ماهانه را می‌توان صفر تعبیر کرد.
# برای آگهی‌های دیگر، مقدار گمشده را حدس نمی‌زنیم.
if {"number of reviews", "reviews per month"}.issubset(df.columns):
    mask_zero_reviews = df["number of reviews"].eq(0) & df["reviews per month"].isna()
    df.loc[mask_zero_reviews, "reviews per month"] = 0

print(f"ردیف‌های کاملاً تکراری حذف‌شده: {exact_duplicates_removed:,}")
print(f"رکوردهای اضافه با شناسه آگهی تکراری حذف‌شده: {duplicate_ids_removed:,}")
print(f"تعداد نهایی ردیف‌ها: {len(df):,}")
display(df.head())
```

## ۳. بررسی کیفیت پس از پاک‌سازی

```python
quality_after = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percent": (df.isna().mean() * 100).round(2),
}).sort_values("missing_percent", ascending=False)
display(quality_after)

fig, ax = plt.subplots(figsize=(11, 5))
plot_missing = quality_after[quality_after["missing_count"] > 0].head(15).sort_values("missing_percent")
ax.barh(plot_missing.index, plot_missing["missing_percent"])
ax.set_title("درصد مقادیر گمشده در ۱۵ ستون دارای بیشترین مقدار گمشده")
ax.set_xlabel("درصد مقادیر گمشده")
ax.set_ylabel("ستون")
plt.tight_layout()
plt.show()
```

### تفسیر کیفیت داده
مقادیر گمشده در همه ستون‌ها یکسان نیستند. برای مثال، `license` ممکن است تقریباً برای همه آگهی‌ها خالی باشد؛ این ستون برای تحلیل‌های عددی اصلی ضروری نیست. در مقابل، گمشدگی `price` مستقیماً روی تعداد رکوردهای قابل استفاده در تحلیل قیمت اثر می‌گذارد. به همین دلیل، در تحلیل‌های بعدی تعداد مشاهدات هر آزمون مشخص می‌شود و داده‌های گمشده به‌صورت کورکورانه با میانگین پر نمی‌شوند.

## ۴. داده‌های پرت قیمت: شناسایی، نه حذف خودکار

```python
price_valid = df["price"].dropna()
q1, q3 = price_valid.quantile([0.25, 0.75])
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
df["price_outlier_iqr"] = df["price"].notna() & ~df["price"].between(lower_fence, upper_fence)

print(f"Q1 قیمت: ${q1:,.2f}")
print(f"Q3 قیمت: ${q3:,.2f}")
print(f"IQR قیمت: ${iqr:,.2f}")
print(f"حد پایین IQR: ${lower_fence:,.2f}")
print(f"حد بالای IQR: ${upper_fence:,.2f}")
print(f"تعداد موارد علامت‌گذاری‌شده به‌عنوان پرت: {int(df['price_outlier_iqr'].sum()):,}")

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
sns.histplot(df["price"].dropna(), bins=40, kde=True, ax=axes[0])
axes[0].set_title("توزیع قیمت (با مقادیر پرت)")
axes[0].set_xlabel("قیمت")
sns.boxplot(x=df["price"], ax=axes[1])
axes[1].set_title("Boxplot قیمت")
axes[1].set_xlabel("قیمت")
plt.tight_layout()
plt.show()
```

## ۵. مقیاس‌دهی (Scaling)

مقیاس‌دهی برای همه تحلیل‌های EDA ضروری نیست؛ نمودار قیمت باید ابتدا در واحد اصلی دلار بررسی شود. در این بخش صرفاً دو روش را روی چند ویژگی عددی نمونه اجرا می‌کنیم تا تفاوت آن‌ها روشن شود. مقیاس‌دهنده‌ها فقط روی ردیف‌های کاملِ ویژگی‌های انتخابی fit می‌شوند.

```python
scale_features = [c for c in ["price", "minimum nights", "number of reviews", "availability 365"] if c in df.columns]
scale_data = df[scale_features].dropna()
print("ویژگی‌های مقیاس‌دهی:", scale_features)
print("تعداد ردیف‌های کامل برای این مثال:", len(scale_data))

if len(scale_data) > 0:
    standard = pd.DataFrame(StandardScaler().fit_transform(scale_data), columns=scale_features, index=scale_data.index)
    minmax = pd.DataFrame(MinMaxScaler().fit_transform(scale_data), columns=scale_features, index=scale_data.index)
    print("آمار قبل از مقیاس‌دهی:")
    display(scale_data.describe().T)
    print("آمار پس از StandardScaler:")
    display(standard.describe().T)
    print("آمار پس از MinMaxScaler:")
    display(minmax.describe().T)

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    sns.boxplot(data=scale_data, ax=axes[0])
    axes[0].set_title("قبل از مقیاس‌دهی")
    sns.boxplot(data=standard, ax=axes[1])
    axes[1].set_title("StandardScaler")
    sns.boxplot(data=minmax, ax=axes[2])
    axes[2].set_title("MinMaxScaler")
    for ax in axes: ax.tick_params(axis="x", rotation=45)
    plt.tight_layout()
    plt.show()
```

## ۶. ذخیره نسخه پاک‌سازی‌شده برای دفترچه دوم

نسخه پاک‌سازی‌شده، همه ستون‌های مفید و پرچم داده پرت را نگه می‌دارد. فایل جداگانه‌ای که فقط قیمت‌های معتبر دارد نیز ساخته می‌شود تا تحلیل قیمت به‌سادگی بازتولید شود. در دفترچه دوم هم فیلترهای لازم صریحاً توضیح داده می‌شوند.

```python
OUT_DIR = Path("../data")
if not OUT_DIR.exists():
    OUT_DIR = Path("data")
OUT_DIR.mkdir(parents=True, exist_ok=True)

clean_path = OUT_DIR / "airbnb_cleaned.csv"
price_path = OUT_DIR / "airbnb_price_analysis.csv"
df.to_csv(clean_path, index=False)
df.loc[df["price"].notna()].to_csv(price_path, index=False)

print("ذخیره شد:", clean_path.resolve())
print("ذخیره شد:", price_path.resolve())
print(f"نسخه پاک‌سازی‌شده: {len(df):,} ردیف")
print(f"رکوردهای دارای قیمت معتبر: {df['price'].notna().sum():,}")
```
